```python
# taking inputs name, age, and city from the user for gold client
def get_client_info():
    name = input("Enter your name: ")
    age = input("Enter your age: ")
    city = input("Enter your city: ")
    
    return name, age, city
```

```python
# manually put the live gold values for currently in the market

gold_price_per_gram = 14826.00  # Example price, replace with actual live value
```

```python
# calculating the total cost of gold based on the user's request
def calculate_gold_cost(weight_in_grams):
    total_cost = weight_in_grams * gold_price_per_gram
    return total_cost
```

```python
# lets call all the functions and get the required output
def main():
    name, age, city = get_client_info()
    weight_in_grams = float(input("Enter the weight of gold in grams: "))
    total_cost = calculate_gold_cost(weight_in_grams)
    
    print(f"\nClient Information:")
    print(f"Name: {name}")
    print(f"Age: {age}")
    print(f"City: {city}")
    print(f"Total cost of {weight_in_grams} grams of gold: {total_cost:.2f} rupees")
```

```python
if __name__ == "__main__":
    main()
    
```
